## 1. Introduction to GitHub & Git Fundamentals {#intro}

### What is GitHub?
- **Platform** for version control and collaborative development hosted on the cloud
- Enables teams to work on **open source** and private projects
- Provides tools for **code hosting, collaboration, and automation**
- Central hub for modern software development workflows

### What is Git?
- **Distributed version control system** (DVCS) that tracks code changes
- Operates **locally** on your machine while allowing remote synchronization
- Stores complete project history; enables branching and merging
- **Free and open-source**; industry standard for version control

### GitHub vs. Git
| Feature | Git | GitHub |
|---------|-----|--------|
| Type | DVCS tool | Cloud-based hosting platform |
| Installation | Local on machine | Cloud-based service |
| Primary Use | Track changes | Collaboration & hosting |
| Requires Internet | No | Yes |
| Cost | Free | Free + Paid plans |

### Key Concepts
- **Repository**: Project folder tracked by Git; contains all project files and history
- **Local Repository**: Git repo on your machine
- **Remote Repository**: Repo hosted on GitHub (origin, upstream)
- **Commit**: Snapshot of changes with message describing what changed
- **Branch**: Independent line of development; main default branch

## 2. Working with GitHub Repositories {#repositories}

### Repository Basics
- **Creating a Repo**: Initialize locally with `git init` or create on GitHub
- **README Files**: Essential documentation; displayed on repo homepage
- **Cloning**: Download complete remote repo with full history (`git clone <url>`)
- **Forking**: Create independent copy of someone else's repo under your account
- **Local vs. Remote**: Local = on your PC; Remote = hosted on GitHub (typically called "origin")

### Repository Structure
- **Root directory**: Contains project files, .git folder, .gitignore, README
- **.git folder**: Hidden directory storing all version history and metadata
- **.gitignore**: File specifying which files Git should ignore (e.g., node_modules, .env)
- **LICENSE**: Specifies how others can use your code (MIT, Apache, GPL, etc.)

### Repository Configuration
- **Public vs. Private**: Public = visible to anyone; Private = only invited collaborators
- **README**: First impression; explains project purpose, setup, and usage
- **Visibility Changes**: Can toggle between public/private in Settings
- **Archive**: Mark repo as read-only (for inactive projects)
- **Delete**: Permanently remove repo (cannot be undone)

### Repository Management Workflows
- **Clone**: `git clone <repo-url>` — Download repo locally
- **Fork**: Create independent version on GitHub; enables open-source contributions
- **Templates**: Use repo templates to quickly create new projects with predefined structure
- **Sync**: Keep forked repo updated with original repository

## 6. Project Management {#projects}

### GitHub Issues for Task Tracking
- **Creation**: Click "New Issue" in repo; title and description required
- **Labels**: Tag issues (bug, enhancement, documentation, help wanted, good first issue)
- **Assignments**: Assign to team members responsible for resolution
- **Milestones**: Group issues into release versions or sprints
- **Linking Issues**: Reference other issues with `#123` syntax

### GitHub Projects
- **Purpose**: Visual task board and workflow management
- **Views**: Table view (spreadsheet-like) and board view (Kanban-style)
- **Automation**: Auto-add issues/PRs linked to project; auto-move between columns
- **Filtering**: View subset of issues by label, assignee, status
- **Templates**: Predefined workflows (To Do → In Progress → Done)

### Release Management
- **Tags**: Mark specific points in history as releases (`git tag v1.0.0`)
- **Releases**: GitHub feature combining tags with release notes and artifacts
- **Semantic Versioning**: Format `MAJOR.MINOR.PATCH` (e.g., 1.2.3)
- **Changelogs**: Document what changed in each release
- **Pre-releases**: Mark beta/RC versions before stable release

### Planning & Roadmaps
- **Milestones**: Organize issues/PRs by target version or date
- **Roadmaps**: High-level overview of planned work (can use Projects or separate tools)
- **Release Planning**: Group related features, bugs, enhancements into versions
- **Sprint Planning**: GitHub Projects can simulate agile sprints

## 9. GitHub Administration {#administration}

### Organization Structure
- **Creating Org**: Settings in personal account under "Organizations" → New organization
- **Organization Roles**: Owner, Member, Outside Collaborator
- **Teams**: Group members for easier permission management and @mentions
- **Access Control**: Base permission + repo-specific overrides
- **Settings**: Visibilty, default permissions, billing contact, member privileges

### Organization Management
- **Inviting Members**: Send invitations; members accept to join org
- **Team Management**: Create teams; assign members; manage team permissions
- **Repository Permissions**: Per-repo access control (read, write, admin)
- **Audit Log**: Track all administrative actions (Enterprise feature)
- **Member Privileges**: Control who can create/delete repos, change visibility, manage security

### Billing & Licensing
- **Free Plan**: Public repos unlimited; private repos limited features
- **Pro Plan**: Focuses on individual developers; unlimited private repos
- **Team Plan**: Shared team workspace; advanced collaboration features
- **Enterprise Plan**: Version control + business features; advanced security
- **Billing Cycle**: Monthly or annual options; manage seat licenses
- **Invoice Access**: Billing history and receipts in Settings → Billing

### Enterprise Features
- **Enterprise Accounts**: Centralized management of multiple organizations
- **Advanced Security**: Secret scanning, dependency management, code scanning
- **Single Sign-On (SSO)**: SAML/OIDC for enterprise identity management
- **IP Allowlisting**: Restrict access by IP address
- **Managed Users**: Provision and manage GitHub users via identity provider
- **Compliance**: Audit logs, export controls, data residency options

## Study Tips for GitHub Foundations Exam Success

### Before the Exam
✅ **Hands-On Practice** (most important!)
- Create a test repo; practice all Git commands locally
- Fork a real open-source repo; submit a test PR
- Create issues, projects, workflows in your practice repo
- Set up SSH keys and test connection
- Configure GitHub Actions with simple workflows

✅ **Command Mastery**
- Master Git commands: `git add`, `commit`, `push`, `pull`, `branch`, `merge`, `reset`
- Understand difference between `git fetch` vs. `git pull`
- Practice resolving merge conflicts manually
- Know `git log` options for troubleshooting

✅ **Concept Review**
- Understand **WHY** not just HOW: Why use branches? Why require code review?
- Know when to use feature vs. Git Flow vs. trunk-based development
- Distinguish between Git (tool) and GitHub (platform)
- Understand authentication trade-offs (SSH vs. PAT)

✅ **Scenario Practice**
- "How would you recover from accidental commit?"
- "Steps to merge someone's PR with conflicts?"
- "Which permissions for team member with limited scope?"
- "How to secure a private repo organization?"

✅ **Documentation Reading**
- Read official GitHub Docs (especially foundations section)
- Review real-world workflows in popular repos
- Check contributing guidelines in 3-5 open-source projects

❌ **Avoid Common Traps**: 
- Confusing `git reset` vs. `git revert`
- SSH keys with wrong permissions (not 600)
- Forgetting 2FA when setting up accounts
- Thinking private repo = no collaboration

✨ **Final Checklist**
- [ ] Can create repo, branch, and push changes locally ✓
- [ ] Can navigate GitHub UI and manage issues/PRs ✓

## 12. Common Exam Focus Areas {#exam-focus}

### Git Scenario Questions
**Scenario**: "You accidentally committed sensitive data. What's the best solution?"
- ❌ Don't use: `git revert` (creates new commit, data still in history)
- ✅ Use: `git reset --soft` (uncommit), or `git rebase --interactive` (rewrite history)
- ✅ Best: Use `BFG Repo-Cleaner` to remove from entire history; force push if private repo

**Scenario**: "Your PR has merge conflicts. How do you resolve?"
- Pull latest main into your branch: `git pull origin main`
- Manually edit conflicted files (keep desired changes)
- Stage resolved files: `git add <file>`
- Complete merge: `git commit -m "Resolve conflicts"`
- Push updated branch; PR auto-updates

### Repository Management
- **When to use Fork**: Contributing to others' repos (no direct write access)
- **When to use Branch**: Own repo feature development (direct write access)
- **Adding Collaborators**: Settings → Collaborators → Invite by username
- **Changing Ownership**: Transfer repo to another user/org (repo admin > settings)
- **Deleting Repo**: Settings → Danger zone (irreversible; 7-day deletion delay)

### Collaboration Best Practices
- **PR Size**: Small, focused PRs merge faster; easier to review
- **Descriptive Titles**: "Add user authentication" vs. "Fix stuff"
- **Meaningful Commits**: One logical change per commit; descriptive messages
- **Review Guidelines**: Check code, tests, docs; comment constructively
- **Status Checks**: PRs should show ✅ all checks passing before approval

### Authentication & Access Questions
- **Choice between SSH and PAT**: SSH for Git ops, PAT for API/GitHub CLI
- **When to rotate tokens**: Regularly (every 90 days); after exposure
- **Multiple SSH keys**: Support multiple keys; useful for different machines/purposes
- **Organization vs. Personal**: Org repos for team; personal repos for individual work
- **Permissions Model**: Start with read; grant write only if needed (least privilege)

### Security & Privacy Decisions
- **Public vs. Private**: Public for open source; Private for commercial/sensitive code
- **Secret Scanning**: GitHub automatically detects exposed secrets; requires rotation
- **Branch Protection**: Always protect main; require reviews and status checks
- **Audit Logs**: Org review logs for compliance; Enterprise feature
- **Data Location**: GitHub has data centers globally; legal/compliance implications

## 10. Modern Development Practices {#devops}

### DevOps & Continuous Integration (CI)
- **CI Definition**: Automatically test code changes as they're committed
- **CI Benefits**: Catch bugs early, maintain code quality, reduce manual testing
- **Build Automation**: Auto-run tests on every push or PR
- **Status Checks**: Show test results; block merging if tests fail

### GitHub Actions Overview
- **Workflow Automation**: Automated tasks triggered by events (push, PR open, schedule)
- **Workflow Files**: YAML configuration in `.github/workflows/` directory
- **Triggers**: `on: [push, pull_request, schedule]` and more
- **Jobs & Steps**: Sequential or parallel execution of tasks
- **Runners**: GitHub-hosted or self-hosted machines execute workflows
- **Use Cases**: Testing, building, deploying, linting, security scanning

### Code Quality Practices
- **Code Review**: Human review before merging catches logic and design issues
- **Branch Protection**: Require PRs and reviews before merging to main
- **Status Checks**: Automated tests must pass; block merge if failing
- **Linting**: Auto-check code style (Prettier, ESLint, flake8, etc.)
- **Testing**: Unit tests, integration tests, end-to-end tests in CI pipeline

### Documentation Best Practices
- **README**: Project overview, setup instructions, usage guidelines
- **Contributing Guidelines**: How to contribute; issue format, PR process
- **Code of Conduct**: Community standards and behavior expectations
- **Wiki**: Supplementary documentation (can be version-controlled or separate)
- **Inline Comments**: Explain complex logic; document why not just what

## 8. Privacy & Permissions {#privacy}

### Repository Visibility
- **Public**: Visible to entire internet; searchable; anyone can fork and clone
- **Private**: Only visible to owner and invited collaborators
- **Internal** (Enterprise): Visible only within enterprise organization
- **Changing Visibility**: Modify in repo Settings → General (requires admin access)

### Access Control & Roles
| Role | Permissions | Use Case |
|------|-------------|----------|
| **Owner** | Full control; delete, transfer repo | Project lead, creator |
| **Collaborator** | Can push/pull/review (configurable) | Team members with write access |
| **Contributor** | Can create PRs from fork | Open source contributors |
| **Member** (Org) | Configurable repo access | Organization team members |
| **Outside Collaborator** | Limited access to specific repos | Contractor, partner access |

### Branch Protection Rules
- **Require PR Reviews**: Minimum number of approvals before merging
- **Require Status Checks**: CI/CD tests must pass before merging
- **Dismiss Stale Reviews**: Old approvals invalid if new commits added
- **Require Admin Review**: Even admins cannot bypass protection on certain branches
- **Lock Main Branch**: Prevent accidental modifications to main

### Secrets Management
- **Repository Secrets**: Store in Settings → Secrets and variables → Repository secrets
- **Environment Secrets**: Secrets specific to deployment environments
- **Secret Types**: Passwords, tokens, API keys, deployment credentials
- **Access**: Only available in GitHub Actions workflows; masked in logs
- **Secret Scanning**: GitHub detects exposed secrets; alerts owner

### Organization Permissions
- **Org Owner**: Full administrative control; manage teams and members
- **Team Maintainer**: Manage specific team; can add/remove members
- **Base Permissions**: Set default permission level for new repos
- **Outside Collaborators**: External people with limited repo access
- **Member Privacy**: Public/private membership visibility in org profile

## 7. Authentication & Security {#auth-security}

### Authentication Methods

| Method | Use Case | Security |
|--------|----------|----------|
| **Username/Password** | Web login; deprecated for API | ❌ Weak, no 2FA protection |
| **Personal Access Token (PAT)** | CLI, Git, API calls | ✅ Strong; scoped permissions |
| **SSH Key Pair** | Git operations without password | ✅ Very strong; asymmetric encryption |
| **OAuth/GitHub Apps** | Third-party integrations | ✅ Secure; granular permissions |
| **SAML/OIDC** | Enterprise SSO | ✅ Enterprise-grade security |

### SSH Setup
- **Key Generation**: `ssh-keygen -t ed25519` (or rsa for older systems)
- **Public Key**: Add to GitHub (Settings → SSH and GPG keys)
- **Private Key**: Keep secure locally; never share
- **Connection**: `ssh -T git@github.com` to test
- **Benefit**: No password needed for Git operations; cryptographically secure

### Personal Access Tokens (PAT)
- **Creation**: Settings → Developer settings → Personal access tokens
- **Scopes**: Choose permissions (repo, workflow, gist, read:org, admin:org_hook, etc.)
- **Expiration**: Set automatic expiration for security (90 days recommended)
- **Rotation**: Regularly create new tokens and retire old ones
- **Storage**: Never commit tokens to repos; use environment variables or secret managers

### Two-Factor Authentication (2FA)
- **Purpose**: Adds second verification step (authenticator app or SMS)
- **Setup**: Settings → Security → Two-factor authentication
- **Recovery Codes**: Save backup codes in secure location; enable account recovery if device lost
- **Enforcement**: Organizations can require 2FA for all members

### Account Security Best Practices
- ✅ Enable 2FA on all accounts
- ✅ Use strong, unique passwords (or password manager)
- ✅ Keep SSH keys secure (chmod 600)
- ✅ Rotate PATs regularly; use environment variables
- ✅ Review Security log regularly (Settings → Security log)
- ✅ Use SSH over HTTPS when possible

## 4. Collaboration Features {#collaboration}

### Pull Requests (PRs)
- **Purpose**: Propose code changes; enables review before merging into main
- **Creating PR**: Create branch → push changes → click "Create Pull Request" on GitHub
- **PR Description**: Explain **what** changed and **why**; link related issues
- **Review Process**: Reviewers leave comments, request changes, or approve
- **Merging Options**: Merge commit, squash, rebase; choose based on project workflow

### Code Review Best Practices
- **Comments**: Ask questions, suggest improvements, identify bugs
- **Suggestions**: Provide specific code improvements reviewers can apply with one click
- **Approvals**: Formal sign-off indicating code is ready to merge
- **Required Reviews**: Enforce minimum reviewers before merging (branch protection)
- **Conversations**: Thread-based discussions on specific code lines

### GitHub Issues
- **Purpose**: Track bugs, feature requests, tasks, and discussions
- **Structure**: Title, description, labels, milestones, assignees
- **Labels**: Categorize issues (bug, feature, documentation, good first issue, etc.)
- **Milestones**: Group issues by release version or deadline
- **Linking**: Reference issues in commits, PRs (`fixes #123` auto-closes issue)

### Notifications & Mentions
- **@mentions**: Tag users or teams to get their attention
- **Notifications**: Receive alerts for watched repos, PRs, issues, mentions
- **Subscription**: "Watch" repo or "Subscribe" to specific conversation threads
- **Email Notifications**: Configure digest vs. real-time notifications

### GitHub Discussions
- **Q&A Format**: Ask questions and get community answers (voted/marked as answer)
- **Announcements**: Share updates with community
- **Ideas**: Propose new features; community feedback before implementation
- **vs. Issues**: Discussions are conversational; Issues are task-focused

## 3. Git Fundamentals & Workflows {#git-workflows}

### Basic Git Concepts
- **Commit**: Snapshot of changes with unique hash; includes message, author, timestamp
- **Staging Area**: Intermediate zone where changes are prepared before committing
- **Working Directory**: Local files on your machine (modified but not staged)
- **Git History**: Complete record of all commits; accessible via `git log`
- **Branches**: Parallel development lines; main default, feature branches for specific work

### Essential Git Commands
| Command | Purpose |
|---------|---------|
| `git add <file>` | Stage changes for commit |
| `git commit -m "message"` | Create commit with staged changes |
| `git push` | Send commits to remote repository |
| `git pull` | Fetch and merge remote changes locally |
| `git fetch` | Download remote changes without merging |
| `git branch` | List, create, delete branches |
| `git checkout`, `git switch` | Change branches |
| `git merge <branch>` | Merge branch into current branch |

### Branching Strategies
- **Feature Branch Workflow**: Create branch for each feature; merge via PR
- **Git Flow**: Separate branches for development, release, hotfix (complex projects)
- **Trunk-Based Development**: Frequent commits to main; minimal long-lived branches
- **GitHub Flow**: One branch (main) + feature branches; simple and effective

### Handling Common Scenarios
- **Merge Conflicts**: Occur when same lines modified differently; resolved manually or via tools
- **Rebase vs. Merge**: Rebase rewrites history (cleaner); Merge preserves history (safer)
- **Git Reset**: Undo commits (`--soft`, `--mixed`, `--hard`); powerful but use carefully
- **Git Revert**: Create new commit undoing previous commit; safer than reset
